```python
#Data viasualisation
"""
Data visualisation involves making a visual representation of the data you're working with.
The most popular tools for visualisation are:
    -matplotlib
    -seaborn
"""
#Lets visualise part of our data...
import matplotlib.pyplot
matplotlib.pyplot.hist(data["Restaurant ID"])

# ==========================================
# CORE TYPES OF DATA VISUALIZATIONS
# ==========================================

# 1. Bar Chart
#    - Purpose: Compares categorical data or distinct groups.
#    - Example: Comparing the average marks across different subjects.

# 2. Line Chart
#    - Purpose: Tracks trends, changes, or developments over time.
#    - Example: Monitoring student performance progression across terms.

# 3. Histogram
#    - Purpose: Shows the distribution and frequency of a single numerical variable.
#    - Example: Visualizing how many students scored in ranges like 60-70 or 80-90.

# 4. Scatter Plot
#    - Purpose: Explores relationships and correlations between two numerical variables.
#    - Example: Checking if study hours correlate with higher final exam marks.

# 5. Box Plot (Whisker Plot)
#    - Purpose: Summarizes data distribution, showing the median, quartiles, and outliers.
#    - Example: Spotting exceptional student scores or overall class performance spread.

# 6. Heatmap
#    - Purpose: Visualizes magnitude or correlation strength using a color-coded grid.
#    - Example: Mapping out a correlation matrix between all subject grades.

# 7. Pie Chart
#    - Purpose: Illustrates percentage proportions or static parts of a whole whole.
#    - Example: Displaying the percentage breakdown of students who passed vs. failed.


#Importance of Visualisation:
#   - Identifies patterns and trends: Helps you quickly spot upward or downward movements over time that are hard to see in raw rows of numbers.
#   - Spots anomalies and outliers: Immediately highlights data entry errors, extreme values, or unusual behavior (like a restaurant with a massive spike in IDs or orders).
#   - Simplifies complex data: Condenses thousands of rows of student marks or restaurant metrics into a single, easy-to-understand image.
#   - Fast-tracks decision making: Allows stakeholders, teachers, or business managers to grasp the core message instantly without reading long reports.
#   - Reveals data distributions: Features like histograms (which you used above) instantly show you how your data is spread out and where most values cluster.
```
